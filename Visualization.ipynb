```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Load your data
csv_file = 'Silicon_Dataset.csv'
df = pd.read_csv(csv_file)

# Ensure all columns are treated as numerical values
df = df.apply(pd.to_numeric)

# Define input and output columns
input_columns = ['Lg', 'Lch', 'Tox', 'Nsub', 'PeakV', 'Vds', 'Vgs']
output_column = 'Id'

# Prepare the data
X = df[input_columns].values
Y = df[output_column].values

# Pick the first 10000 samples
sample_num = 10000
X = X[:sample_num]
Y = Y[:sample_num]

# Extract Vgs for x-axis
Vgs = df['Vgs'][:sample_num]

# Take the logarithm of Id
Y_log = -np.log(Y)

# Plot the graph
plt.figure(figsize=(8,6))
plt.scatter(Vgs, Y_log, s=1, alpha=0.5, label='Log(Id) vs Vgs')
plt.xlabel('Vgs (Gate Voltage)')
plt.ylabel('Log(Id) (Log Drain Current)')
plt.title('Transfer Characteristics: Log(Id) vs Vgs')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

# Define the models and their corresponding MAE values
models = ['MLE', 'RF', 'XGBoost', 'FatGCN (MSE)']
mae_values = [0.011593, 0.007704, 0.006866, 0.006352620664983988]

# Create the bar chart
plt.figure(figsize=(8, 6))
plt.bar(models, mae_values, color=['blue', 'green', 'orange', 'purple'])

# Add labels and title
plt.xlabel('Models')
plt.ylabel('Mean Absolute Error (MAE)')
plt.title('MAE Comparison Between Models')
plt.ylim(0, 0.012)  # Set y-axis limit for better visualization

# Annotate the bars with MAE values
for i, value in enumerate(mae_values):
    plt.text(i, value + 0.0002, f'{value:.6f}', ha='center', fontsize=10)

# Show the grid and the plot
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
# Define the models and their corresponding trainable parameters
models = ['FatGCN', 'MLE']
parameters = [762369, 561]  # Number of trainable parameters

# Create the bar chart
plt.figure(figsize=(4, 6))
plt.bar(models, parameters, color=['purple', 'orange'])

# Add labels and title
plt.xlabel('Models', fontsize=12)
plt.ylabel('Number of Trainable Parameters', fontsize=12)
plt.title('Comparison of Trainable Parameters Between Models', fontsize=14)

# Annotate the bars with parameter values
for i, value in enumerate(parameters):
    plt.text(i, value + 20000, f'{value:,}', ha='center', fontsize=10)

# Add gridlines
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Show the plot
plt.tight_layout()
plt.show()
```
